```python
class General_Newton_Raphson():
    def __init__(self, initial_guess, iteration_num, tolerance, h=0.00001):
        self.iteration_num = iteration_num
        self.tolerance = tolerance
        self.initial_guess = initial_guess
        self.h = h

    def solve(self, f):
        x_n = self.initial_guess
    
        for _ in range(self.iteration_num):
    
            f_val = f(x_n) # ham fonksiyonu alıp içine istediğim değeri yerleştirebildim.

            # nümerik türev (Irr sınıfından türev ifadesini ayırmak için bir çözüm.)
            df_val = (f(x_n+self.h) - f(x_n)) / self.h
    
            if df_val == 0:
                return None
            
            x_next = x_n - (f_val / df_val)
            
            diff = x_next - x_n
            if abs(diff) <= self.tolerance:
                return x_next
            x_n = x_next
            
        return x_n

class Irr():
    def __init__(self, initial_investment, cash_flows):
        self.flows = [initial_investment] + cash_flows

    def calculate_irr(self, engine):

        def npv_f(i):
            return sum(cf / (1+i)**t for t, cf in enumerate(self.flows))

       
        return engine.solve(npv_f) # npv_f fonksiyonunu ham şekilde yazdım ve bu haliyle gönderdim.
        
        
    
```

```python
irr1 = Irr(-80, [30, 30 ,30 ,30 ,30 ,30 ,30 ,30 ,30, 30])
```

```python
newton_raphson = General_Newton_Raphson(0.1, 100, 0.00001)

irr1.calculate_irr(newton_raphson)
```
